# Section 3. Data Quality Assessment

In this section, we assess the quality of the raw dataset before applying any cleaning transformations.

We will check:
- Missing values
- Duplicate records
- Unique values
- Potential data quality issues

```python
# Check missing values
df_raw.isnull().sum()
```

```python
# Check duplicate records
df_raw.duplicated().sum()
```

```python
# Check unique values in categorical columns
categorical_cols = df_raw.select_dtypes(include="object").columns

for col in categorical_cols:
    print(f"\n{col}:")
    print(df_raw[col].value_counts())
```

```python
# Count "Unknown" values in categorical columns
for col in categorical_cols:
    unknown_count = (df_raw[col] == "Unknown").sum()
    if unknown_count > 0:
        print(col, ":", unknown_count)
```

```python

# Identify unnecessary Naive Bayes columns
naive_bayes_cols = [
    col for col in df_raw.columns
    if col.startswith("Naive_Bayes_Classifier")
]

naive_bayes_cols
```

---

# Section 4. Data Cleaning

In this section, the dataset is cleaned by removing unnecessary columns and preparing categorical values for further analysis.

```python
# Create a copy of the raw dataset for cleaning
df_clean = df_raw.copy()

df_clean.shape
```

```python
# Remove unnecessary Naive Bayes classifier columns
df_clean = df_clean.drop(columns=naive_bayes_cols)

# Check the new dataset shape
df_clean.shape
```

### Handling "Unknown" Categories

The categorical variables Education_Level, Marital_Status, and Income_Category contain "Unknown" values.

These records are retained rather than removed to avoid unnecessary data loss. The "Unknown" category may also provide useful information during exploratory data analysis.

```python
# Confirm "Unknown" values are retained
for col in ["Education_Level", "Marital_Status", "Income_Category"]:
    print(col, ":", (df_clean[col] == "Unknown").sum())
```

### Customer Identifier

CLIENTNUM is a unique customer identifier. It is retained in the cleaned dataset for record identification but will not be used as a predictive feature in the machine learning model.


```python
# Check whether CLIENTNUM uniquely identifies each customer
print("Total rows:", len(df_clean))
print("Unique CLIENTNUM:", df_clean["CLIENTNUM"].nunique())
```

### Create Churn Target Variable

A binary Churn variable is created from Attrition_Flag to simplify analysis and prepare the dataset for machine learning.

- 0 = Existing Customer
- 1 = Attrited Customer


```python
# Create binary churn target
df_clean["Churn"] = df_clean["Attrition_Flag"].map({
    "Existing Customer": 0,
    "Attrited Customer": 1
})

# Check the result
df_clean[["Attrition_Flag", "Churn"]].value_counts()
```

---

# Section 5. Prepare Dataset for Analysis

The cleaned dataset is prepared for exploratory data analysis and subsequent machine learning.


```python
# Remove the original target column
df_processed = df_clean.drop(columns=["Attrition_Flag"])

df_processed.shape
```
